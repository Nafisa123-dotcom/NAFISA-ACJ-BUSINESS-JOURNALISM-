## 4. Cleaning the Dataset

The original Excel sheet contains the main district-level voting data as well as a separate Nagpur voting update.

For this analysis, we focus first on the district-level voting percentages for the 2009, 2014, 2019 and 2024 Lok Sabha elections.

```python
main = df.iloc[1:12, 0:7].copy()
```

```python
main.columns = [
    "Centre",
    "Sr_No",
    "District",
    "2024",
    "2019",
    "2014",
    "2009"
]
```

### Cleaned Dataset

## 5. Converting Voting Figures into Percentages

The voting figures are stored as decimal values. We convert them into percentages so that they are easier to interpret and compare.

```python
for year in ["2024", "2019", "2014", "2009"]:
    main[year] = main[year] * 100
```

```python
main
```

## 7. Highest and Lowest Voting Percentages

We identify the districts with the highest and lowest voting percentages in each election year. This helps us understand the differences in voter turnout across districts.

```python
for year in ["2024", "2019", "2014", "2009"]:
    highest = main.loc[main[year].idxmax(), ["District", year]]
    lowest = main.loc[main[year].idxmin(), ["District", year]]
    
    print(year)
    print("Highest:", highest.to_dict())
    print("Lowest:", lowest.to_dict())
    print()
```

# 12. Key Findings

The analysis shows that voter turnout in the districts covered by the dataset increased over the long term.

- The average voting percentage increased from approximately 63.64% in 2009 to 68.05% in 2024.
- This represents an increase of approximately 4.42 percentage points.
- All 11 districts recorded higher turnout in 2024 than in 2009.
- All 11 districts also recorded higher turnout in 2024 than in 2019.
- Akola recorded the largest increase between 2019 and 2024, rising from 58.36% to 64.98%, an increase of 6.62 percentage points.

These findings suggest that the rise in voter turnout in 2024 was widespread across the districts in the dataset rather than being limited to a small number of districts.

# 13. Story Pitch

## Vidarbha Sees Broad Rise in Voter Turnout

Voter turnout across the Vidarbha districts covered in the dataset increased in the 2024 Lok Sabha election, with all 11 districts recording higher participation than in 2009 and 2019. Average turnout rose from 63.64% in 2009 to 68.05% in 2024, an increase of 4.42 percentage points. Akola recorded the biggest increase between 2019 and 2024, with turnout rising by 6.62 percentage points to 64.98%.

The data raises a broader reporting question: what drove the increase in voter participation across Vidarbha in 2024? A follow-up story could examine whether the rise was linked to voter-awareness campaigns, changes in the electorate, local political issues, or differences in urban and rural participation. Interviews with election officials, political analysts and voters could help explain the reasons behind the trend.
